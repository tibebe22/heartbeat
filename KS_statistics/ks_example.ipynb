```python
import numpy as np
from scipy.stats import ks_2samp
#load data
nsr = np.loadtxt('example_nsr.txt')
chf = np.loadtxt('example_chf.txt')
af = np.loadtxt('example_af.txt')

#calculate the K-S test
ks_nsr_chf = ks_2samp(nsr,chf)
ks_nsr_af = ks_2samp(nsr,af)
ks_chf_af = ks_2samp(chf,af)

#open the file
with open('ks_example.txt','w') as f:
    f.write(f"ks test nsr vs chf: statstic = {ks_nsr_chf.statistic}, p-value = {ks_nsr_chf.pvalue}\n")
    f.write(f"ks test nsr vs af: statstic = {ks_nsr_af.statistic}, p-value = {ks_nsr_af.pvalue}\n")
    f.write(f"ks test chf vs af: statstic = {ks_chf_af.statistic}, p-value = {ks_chf_af.pvalue}\n")
```
